```python
import numpy as np
department = ['IT', 'HR', 'Sales', 'Finance']
male = [15, 8, 18, 12]
female = [10, 7, 12, 8]
x = np.arange(len(department))
plt.bar(x,male,width=0.4, label='Male')
plt.bar(x + 0.4, female, width=0.4, label='Female')
plt.xticks(x + 0.2, department)
plt.xlabel('Department')
plt.ylabel('Employees')
plt.title('Employees by Department')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

department = ['IT', 'HR', 'Sales', 'Finance']
male = [15, 8, 18, 12]
female = [10, 7, 12, 8]
x = np.arange(len(department))
plt.bar(x,male,width=0.4,label='Male')
plt.bar(x+0.4,female,width=0.4,label='Female')
plt.xticks(x+0.2,department)
plt.xlabel('Department')
plt.ylabel('Employees')
plt.title('Employees by Department')
plt.legend()
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [20000, 25000, 22000, 30000]
plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
plt.plot(months,sales)
plt.title('Sales Trends')

plt.subplot(2, 2, 2)
plt.bar(months, sales)
plt.title('Sales by Month')

plt.subplot(2, 2, 3)
plt.plot(months, sales)
plt.title('Sales Trends2')

plt.subplot(2, 2, 4)
plt.bar(months, sales)
plt.title('Sales by Month 2')
plt.tight_layout()
plt.show()
```
